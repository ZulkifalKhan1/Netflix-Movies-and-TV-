```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='darkgrid')
plt.rcParams['figure.figsize'] = (10, 5)

df = pd.read_csv('../data/netflix_titles.csv')
print(df.shape)
df.head()
```

```python
# ── Basic Info ──────────────────────────────────────────────────────────────
print('Columns:', df.columns.tolist())
print('\nDtypes:\n', df.dtypes)
print('\nMissing values:\n', df.isnull().sum())
```

```python
# ── Descriptive Statistics ──────────────────────────────────────────────────
df.describe(include='all')
```

```python
# ── Data Cleaning ───────────────────────────────────────────────────────────
df.drop_duplicates(subset=['title','type','release_year'], keep='first', inplace=True)
df.fillna({'director':'Unknown','cast':'Unknown','country':'Unknown','rating':'Not Rated'}, inplace=True)

def parse_duration(row):
    val = str(row['duration'])
    if 'min' in val:
        return int(val.replace('min','').strip())
    elif 'Season' in val:
        return int(val.replace('Seasons','').replace('Season','').strip()) * 10
    return 0

df['duration_min'] = df.apply(parse_duration, axis=1)
df['country']      = df['country'].str.split(',').str[0].str.strip()
df['date_added']   = pd.to_datetime(df['date_added'], errors='coerce')
df['year_added']   = df['date_added'].dt.year
print('After cleaning:', df.shape)
```

```python
# ── Chart 1: Pie ─────────────────────────────────────────────────────────────
counts = df['type'].value_counts()
plt.figure()
plt.pie(counts, labels=counts.index, autopct='%1.1f%%', colors=['#E50914','#444'])
plt.title('Movies vs TV Shows')
plt.tight_layout()
plt.show()
```

```python
# ── Chart 2: Histogram ───────────────────────────────────────────────────────
movies = df[df['type']=='Movie']['duration_min']
movies = movies[(movies>0) & (movies<300)]
plt.figure()
plt.hist(movies, bins=30, color='#E50914', edgecolor='black', alpha=0.8)
plt.axvline(movies.mean(), color='yellow', linestyle='--', label=f'Mean: {movies.mean():.0f} min')
plt.xlabel('Duration (min)'); plt.ylabel('Count'); plt.title('Movie Duration Distribution')
plt.legend(); plt.tight_layout(); plt.show()
```

```python
# ── Chart 3: Line ────────────────────────────────────────────────────────────
yearly = df.groupby(['release_year','type']).size().unstack(fill_value=0)
yearly[yearly.index >= 1990].plot(color=['#E50914','#4FC3F7'])
plt.title('Content Released per Year'); plt.xlabel('Year'); plt.ylabel('Count')
plt.tight_layout(); plt.show()
```

```python
# ── Chart 4: Bar ─────────────────────────────────────────────────────────────
top = df['country'].value_counts().head(10)
top.sort_values().plot(kind='barh', color='#E50914')
plt.title('Top 10 Countries by Content'); plt.xlabel('Count')
plt.tight_layout(); plt.show()
```

```python
# ── Chart 5: Scatter ─────────────────────────────────────────────────────────
mv = df[(df['type']=='Movie') & (df['duration_min']>0)]
plt.scatter(mv['release_year'], mv['duration_min'], alpha=0.3, color='#E50914', s=10)
plt.xlabel('Release Year'); plt.ylabel('Duration (min)')
plt.title('Release Year vs Movie Duration'); plt.tight_layout(); plt.show()
```
